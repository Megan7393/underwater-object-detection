```python
from pathlib import Path  
import pandas as pd

PROJECT_ROOT = Path(
    r"C:\Users\megdo\Desktop\underwater-object-detection"
)

COMPARISON_DIR = PROJECT_ROOT / "results" / "model_comparison"
COMPARISON_DIR.mkdir(
    parents=True,
    exist_ok=True,
)

model_comparison = pd.DataFrame(
    [
        {
            "model": "YOLOv8n",
            "mAP_50": 0.6542,
            "mAP_50_95": 0.4456,
            "inference_ms_per_image": 18.3,
            "fps": 54.6,
        },
        {
            "model": "Faster R-CNN ResNet50-FPN",
            "mAP_50": 0.6949,
            "mAP_50_95": 0.4721,
            "inference_ms_per_image": 530.85,
            "fps": 1.88,
        },
        {
            "model": "DETR ResNet-50",
            "mAP_50": 0.4659,
            "mAP_50_95": 0.2359,
            "inference_ms_per_image": 146.26,
            "fps": 6.84,
        },
    ]
)

model_comparison = model_comparison.sort_values(
    by="mAP_50_95",
    ascending=False,
).reset_index(drop=True)

comparison_path = (
    COMPARISON_DIR / "baseline_model_comparison.csv"
)

model_comparison.to_csv(
    comparison_path,
    index=False,
)

display(model_comparison)

print("Saved to:", comparison_path)
```

```python
#per class ap comparison
per_class_comparison = pd.DataFrame(
    [
        {
            "class": "holothurian",
            "YOLOv8n_AP_50_95": 0.3879,
            "Faster_R_CNN_AP_50_95": 0.4561,
            "DETR_AP_50_95": 0.2180,
        },
        {
            "class": "echinus",
            "YOLOv8n_AP_50_95": 0.6158,
            "Faster_R_CNN_AP_50_95": 0.6190,
            "DETR_AP_50_95": 0.3618,
        },
        {
            "class": "scallop",
            "YOLOv8n_AP_50_95": 0.2137,
            "Faster_R_CNN_AP_50_95": 0.2695,
            "DETR_AP_50_95": 0.0578,
        },
        {
            "class": "starfish",
            "YOLOv8n_AP_50_95": 0.5649,
            "Faster_R_CNN_AP_50_95": 0.5437,
            "DETR_AP_50_95": 0.3057,
        },
    ]
)

per_class_path = (
    COMPARISON_DIR / "per_class_model_comparison.csv"
)

per_class_comparison.to_csv(
    per_class_path,
    index=False,
)

display(per_class_comparison)

print("Saved to:", per_class_path)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.bar(
    model_comparison["model"],
    model_comparison["mAP_50_95"],
)

plt.title("Baseline Model Comparison on the DUO Test Set")
plt.xlabel("Model")
plt.ylabel("mAP@0.5:0.95")
plt.ylim(0, 0.55)
plt.xticks(rotation=15)
plt.tight_layout()

accuracy_figure_path = (
    COMPARISON_DIR / "baseline_map_comparison.png"
)

plt.savefig(
    accuracy_figure_path,
    dpi=300,
    bbox_inches="tight",
)

plt.show()

print("Saved to:", accuracy_figure_path)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    model_comparison["model"],
    model_comparison["fps"],
)

plt.title("Baseline Model Inference Speed on the DUO Test Set")
plt.xlabel("Model")
plt.ylabel("Frames per second")
plt.xticks(rotation=15)
plt.tight_layout()

speed_figure_path = (
    COMPARISON_DIR / "baseline_fps_comparison.png"
)

plt.savefig(
    speed_figure_path,
    dpi=300,
    bbox_inches="tight",
)

plt.show()

print("Saved to:", speed_figure_path)
```
